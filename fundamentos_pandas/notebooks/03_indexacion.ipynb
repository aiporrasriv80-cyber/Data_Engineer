LOC

Localiza las filas por indice, dado que las columnas no tienen indice se debe usar su etiqueta, nombre o listarlas si las queremos en conjunto.

```python
# Busqueda de registros indexando con "loc" y de columnas listando con "[]"
ventas_df.loc[0:4,["vendedor", "producto", "categoria"]]
```

```python
# Del indice 1000 hacia el final, solo se mostrara las columnas "vendedor" y "producto"
ventas_df.loc[1000:,["vendedor", "producto"]]
```

ILOC

Localiza como el slicing similar a python nativo, similar a las listas

```python
ventas_df.iloc[0:4]
```

```python
# Indexando con "iloc" las filas, pero en el caso de las columnas no se lista, sino que se indexa con el numero de columna, en este caso 0 y 1
ventas_df.iloc[0:4, 0:2]
```
