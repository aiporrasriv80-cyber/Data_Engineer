SERIES Y DATAFRAME

```python
# Una serie es un array 1D de datos con etiquetas, es decir, un array indexado.
serie = pd.Series(
    [10, 20, 30],
    index = ['a', 'b', 'c'],
    name = "ventas"
)

# un dataframe es un array 2D o tabn un conjunto de series
df = pd.DataFrame({
    "producto" : ["A", "B"],
    "unidades" : [10, 20]
})
```

DataFrames

Crear DataFrame desde un diccionario

```python
df = pd.DataFrame({
    "producto" : ["Laptop", "Monitor", "Teclado"],
    "unidades" : [3, 7, 12],
    "precio" : [1200, 320, 95]
})

df
```

Desde una lista de diccionarios

```python
df2 = pd.DataFrame(
    [
        {"producto" : "Laptop", "unidades" : 3, "precio" : 1200},
        {"producto" : "Monitor", "unidades" : 7, "precio" : 320}
    ]
)

df2
```

Desde una lista de listas

```python
df3 = pd.DataFrame(
    [
        ["Laptop", 3, 1200],
        ["Monitor", 7, 320]
    ],
    columns=["producto", "unidades", "precio"]
)

df3
```

Columnas Calculadas

```python
df = pd.DataFrame({
    "producto" : ["Laptop", "Monitor", "Teclado"],
    "unidades" : [3, 7, 12],
    "precio" : [1200, 320, 95]
})

df
```

```python
df["importe"] = df["unidades"] * df["precio"]
df
```
