```python
import requests
import pandas as pd

respuesta = requests.get(
    "https://api.open-meteo.com/v1/forecast",
    params={
        "latitude": -16.40,
        "longitude": -71.54,
        "hourly": "temperature_2m,relative_humidity_2m,precipitation",
        "timezone": "America/Lima",
        "forecast_days": 7,
    },
    timeout=30,
)
respuesta.raise_for_status()
```

```python
respuesta.json()
```

```python
datos = respuesta.json()
df = pd.DataFrame(datos["hourly"])
df["time"] = pd.to_datetime(df["time"])

df.head(10)
```

```python
import requests
import pandas as pd

url = "https://pokeapi.co/api/v2/pokemon"

respuesta = requests.get(url, params={"limit": 10}, timeout=30)
respuesta.raise_for_status()
```

```python
respuesta.json()
```

```python
datos = respuesta.json()
df = pd.DataFrame(datos["results"])
df.head()
```

```python
respuesta = requests.get("https://pokeapi.co/api/v2/pokemon/1", timeout=30)
respuesta.raise_for_status()
respuesta.json()
```

```python
filas = []

for url_pokemon in df["url"]:
    respuesta = requests.get(url_pokemon, timeout=30)
    respuesta.raise_for_status()
    pokemon = respuesta.json()

    filas.append({
        "id": pokemon["id"],
        "nombre": pokemon["name"],
        "altura_m": pokemon["height"] / 10,
        "peso_kg": pokemon["weight"] / 10,
        "tipos": ", ".join(
            elemento["type"]["name"]
            for elemento in pokemon["types"]
        ),
    })

df_pokemon = pd.DataFrame(filas)
df_pokemon.head(50)
```
