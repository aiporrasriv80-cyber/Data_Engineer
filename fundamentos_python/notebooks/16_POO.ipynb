Programación Orientada a Objetos (POO)

```python

class Perro:
    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad

    def ladrar(self):
        return f"{self.nombre} dice: Guau!"

    def cumplir_anios(self):
        self.edad += 1
```

```python
fido = Perro("Fido", 5)
```

```python
print(fido.ladrar())
```
