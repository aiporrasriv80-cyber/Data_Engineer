```python
import io

path = "../data/notas.txt"

lista = ["Primera Linea", "Segunda Linea", "Tercera Linea", "Cuarta Linea", "Quinta Linea"]

def write(f: io.TextIOWrapper, linea: list[str]) -> None:
    f.writelines([l + "\n" for l in linea])

with open(path, "w") as f:
    write(f, lista)
```

```python
with open(path, "a") as f:
    f.write("Quinta Linea")
```

```python
### Guardar utilizando salto línea automático
with open(path, "w") as f:
    print(*lista, sep="\n", file=f)
```

```python
from datetime import datetime
import time

total_intentos = 3
n_intento = 0
errores = []
path = "../data/ventas_20260926.txt"

while n_intento < total_intentos:
    try:
        with open(path, "r") as f:
            contenido = f.read()
            print(f"El contenido es: {contenido}")
            break
    except FileNotFoundError as e:
        errores.append(f"{datetime.now()} | Error: {e} | Número de intento: {n_intento + 1}")
        n_intento += 1
        time.sleep(10)

print(errores)
```
