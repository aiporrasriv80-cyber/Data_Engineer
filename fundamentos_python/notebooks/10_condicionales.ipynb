```python
color = "verde"

if color == "verde":
    premio = 10
elif color == "azul":
    premio = 5
else:
    premio = 0

print(premio)
```

```python
color = "rosado"

if color == "verde":
    premio = 10
elif color == "azul":
    premio = 5
elif color == "rosado":
    premio = 3.5
elif color == "negro":
    premio = 1
else:
    premio = 0

print(premio)
```

Expresiones condicionales

```python
registros = []

print("Llenar registros de hoy dia...")

if registros:
    print("Procesando...")
else:
    print("Los registros están vacíos")
```

```python
registros = []

print("Llenar registros de hoy dia...")
registros.extend([1, 2, 3, 4, 5])

if registros:
    print("Procesando...")
else:
    print("Los registros están vacíos")
```

```python
### NADA PYTHONISTA, NO SE DEBERÍA PARA TENER UN CÓDIGO MÁS LIMPIO

cantidad_registros = len(registros)

if cantidad_registros > 0:
    print("Procesando...")
```
