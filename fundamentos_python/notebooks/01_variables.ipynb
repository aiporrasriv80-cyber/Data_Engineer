TIPO DE VARIABLES

```python
cantidad = 3
precio = 145.0
cliente = "Marco Torres"
activo = True
descuento = None
```

```python
cantidad: int = 3
precio: float = 145.0
cliente: str = "Marco Torres"
activo: bool = True
descuento: None = None
```

```python
print("Tipo de la variable cantidad:",type(cantidad))
print(type(precio))
print(type(descuento))
```

CASTEO DE VARIABLES

```python
cantidad = "3"
print(type(cantidad))

cantidad = int(cantidad)
print(type(cantidad))
```

```python
precio = "130.5"
print(type(precio))

precio = float(precio)
print(type(precio))
print(precio)

precio = int(precio)
print(precio)
```

```python
activo = 0
print(activo)
activo = bool(activo)
print(activo)
```

```python
activo = -1
print(activo)
activo = bool(activo)
print(activo)
```
